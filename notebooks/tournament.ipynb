```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

df = pd.read_csv("../tournaments/tournament_results.csv")
df["X_win_rate"] = df["X_wins"] / df["games"]

pivot = df.pivot(index="X_bot", columns="O_bot", values="X_win_rate")

sns.heatmap(pivot, annot=True, cmap="Blues", vmin=0, vmax=1)
plt.title("X Win Rate by Matchup")
plt.show()
```
